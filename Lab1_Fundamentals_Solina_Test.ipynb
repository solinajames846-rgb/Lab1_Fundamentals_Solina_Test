```python
import os 
import subprocess
print("=== 1. EXECUTION LOG ===")
os.makedirs("conflict_simulation", exist_ok=True)
os.chdir("conflict_simulation")
commands = [
    "git init",
    'echo print("Original Line") > main.py',
    "git add main.py",
    "git commit -m 'Initial commit'"
    'echo print("Edited by main branch") > main.py',
    "git add main.py",
    'git commit -m "Main updates main.py"',
    "git checkout simulation",
    'echo print("Edited by simulation branch") > main.py',
    "git add main.py",
    'git commit -m "Simulation updates main.py"',
    "git checkout main",
    "git merge simulation"

]
for cmd in commands:
    print(f"Running: {cmd}")
    result = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    if result.stdout: print(result.stdout.strip())
    if result.stderr: print(result.stderr.strip())
print
("\n=== 2. SYSTEM ARTIFACT (CONFLICT TEXT) ===")
if os.path.exists("main.py"):
    with open("main.py", "r") as f:
        with open("main.py", "r") as f:
            content = f.read()
        print("Content of main.py showing conflict markers:\n")
        print(content)
```
